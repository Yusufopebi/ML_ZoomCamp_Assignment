```python
# Select all the cars from Asia, select only columns vehicle_weight and model_year
asia_cars = df[df["origin"] == "Asia"][["vehicle_weight", "model_year"]]
asia_cars
```

```python

#Q7.3 Select the first 7 values (but there are only 5 rows and none are Asia)
X = asia_cars.head(7).to_numpy()

# X is empty because there are no Asia cars
print("X is empty:", X)
```

```python
# Select rows where origin is "Asia"
asia_rows = df[df["origin"] == "Asia"]

print(asia_rows)
```

```python
# Compute matrix-matrix multiplication between the transpose of X and X
if X.size == 0:
    XTX = np.array([])  # or handle as needed for empty array
else:
    XTX = X.T @ X
```

```python

# Invert XTX
if XTX.size == 0:
 XTX_inv = np.array([])  # Handling for empty
else:
    try:
        XTX_inv = np.linalg.inv(XTX)
    except np.linalg.LinAlgError:
        XTX_inv = np.array([])  # Handle singular matrix
```

```python


# Create an array y
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python

# Multiply the inverse of XTX with the transpose of X, and then multiply the result by y
if X.size == 0 or XTX_inv.size == 0:
    w = np.array([])  # Because X is empty
else:
    w = XTX_inv @ X.T @ y
```

```python

# Sum of all elements of w
if w.size == 0:
    sum_w = 0
else:
    sum_w = np.sum(w)

print("Sum of w:", sum_w)
```
